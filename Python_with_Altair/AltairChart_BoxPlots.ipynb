```python
import pandas as pd
import altair as alt 
```

```python
auction_data = pd.read_csv('datasets/auction.csv')
```

```python
auction_data.head(5)
```

```python
auction_data.shape
```

```python
# Reduces data to fewer rows, 2500 to be exact
auction_data = auction_data.sample(2500)
```

```python
auction_data.shape
```

```python
# Track unique items
auction_data['item'].unique()
```

```python
# Extracting a unique data type
auction_xbox = auction_data[auction_data['item'] == 'Xbox game console']
```

```python
auction_xbox.shape
```

```python
alt.Chart(auction_xbox, height=400, width=600) \
    .mark_boxplot() \
    .encode(x='item', y='bid')
```

```python
# Specifying the box size
alt.Chart(auction_xbox, height=400, width=600) \
    .mark_boxplot(size=100) \
    .encode(x='item', y='bid')
```

```python
alt.Chart(auction_data, height=400, width=600) \
    .mark_boxplot(size=80) \
    .encode(x='item', y='bid')
```

```python
alt.Chart(auction_xbox, height=600, width=600) \
    .mark_boxplot(size=100) \
    .encode(x='item', y='bid', color='auction_type')
```

```python
auction_watch = auction_data[auction_data['item'] == 'Cartier wristwatch']
```

```python
alt.Chart(auction_watch, height=400, width=600) \
    .mark_boxplot(size=100) \
    .encode(x='item', y='bid', color='auction_type')
```

```python
# To see more numeric values
alt.Chart(auction_watch, height=400, width=600) \
    .mark_boxplot(size=100) \
    .encode(x='item', y='price', color='bid')
```

```python
# Turn chart 90 Degrees by swapping x and y
alt.Chart(auction_watch, height=400, width=600) \
    .mark_boxplot(size=100) \
    .encode(x='price', y='item', color='bid')
```

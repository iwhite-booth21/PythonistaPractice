```python
import altair as alt
import pandas as pd
```

```python
housing_data = pd.read_csv("datasets/Melbourne_housing.csv")
```

```python
# Lets get a feel for the data we have in the file
housing_data.head()
```

```python
housing_data.shape
```

```python
#Remove entries with N/A
housing_data = housing_data.dropna()
```

```python
housing_data.shape
```

```python
# Determine the unique datatypes
housing_data['Type'].unique()
```

```python
# Modify the data to capture data of type 'T'
housing_data = housing_data[housing_data['Type']=='t']
```

```python
housing_data.shape
```

```python
# Chart out the newly proportioned data:
alt.Chart(housing_data, height=400, width=600).mark_circle().encode(
    x='Distance',
    y='Price'
)
```

```python
# Keep chart, change x :
alt.Chart(housing_data, height=400, width=600).mark_circle().encode(
    x='BuildingArea',
    y='Price'
)
```

```python
# Specify the building area
housing_data = housing_data[housing_data['BuildingArea']>=30]
```

```python
# Using new build area :
alt.Chart(housing_data, height=400, width=600).mark_circle().encode(
    x='BuildingArea',
    y='Price'
)
```

```python
# Specify the size:
alt.Chart(housing_data, height=400, width=600).mark_circle(size=60).encode(
    x='BuildingArea',
    y='Price'
)
```

```python
# Change the y:
alt.Chart(housing_data, height=400, width=600).mark_circle(size=60).encode(
    x='BuildingArea',
    y='Landsize'
)
```

```python
# Set access limits on the y:
alt.Chart(housing_data, height=400, width=600).mark_circle(size=60).encode(
    x='BuildingArea',
    y=alt.Y('Landsize', scale=alt.Scale(domain=(0,1500))),
)
```

```python
# add in clip to ensure data stays within chart

alt.Chart(
    housing_data,
    height=400,
    width=600
).mark_circle(
    size=60,
    clip=True
).encode(
    x='BuildingArea',
    y=alt.Y('Landsize', scale=alt.Scale(domain = (0,1500))),
)
```

```python
# Checking the rooms price
alt.Chart(housing_data, width=600, height=400).mark_circle(size=60).encode(
    x='Rooms',
    y='Price'
)
```

```python
# Using the color to check the value of rooms while checking the Building Area information

alt.Chart(housing_data, width=600, height=400).mark_circle(size=60).encode(
    x='BuildingArea:Q',
    y='Price:Q',
    color='Rooms:O'
)
```

```python
# Set Rooms to be the nominal value

alt.Chart(housing_data, width=600, height=400).mark_circle(size=60).encode(
    x='BuildingArea:Q',
    y='Price:Q',
    color='Rooms:N'
)
```

```python
# Make chart interactive

alt.Chart(housing_data, width=600, height=400).mark_circle(size=60).encode(
    x='BuildingArea:Q',
    y='Price:Q',
    color='Rooms:N'
).interactive()
```

```python
# Lets customize scatterplots
alt.Chart(housing_data, width=600, height=400).mark_circle(size=60).encode(
    x = 'BuildingArea:Q',
    y='Price:Q',
    color='Regionname:N'
).interactive()
```

```python
# Lets customize scatterplots - Adding size
alt.Chart(housing_data, width=600, height=400).mark_circle(size=60).encode(
    x = 'BuildingArea:Q',
    y='Price:Q',
    color='Regionname:N',
    size='Distance:Q'
).interactive()
```

```python
# Lets customize scatterplots - adding size and x:bins
alt.Chart(housing_data, width=600, height=400).mark_circle(size=60).encode(
    x = alt.X('BuildingArea:Q',bin=True),
    y='Price:Q',
    color='Regionname:N',
    size='Distance:Q'
)
```

```python
# Lets customize scatterplots - adding size and y:bins
alt.Chart(housing_data, width=600, height=400).mark_circle(size=60).encode(
    x = 'BuildingArea:Q',
    y=alt.Y('Price:Q', bin=True),
    color='Regionname:N',
    size='Distance:Q'
)
```

```python
# Lets customize using mark_point - Adding size
alt.Chart(housing_data, width=600, height=400).mark_point(size=60).encode(
    x = 'BuildingArea:Q',
    y='Price:Q',
    color='Regionname:N',
    size='Distance:Q'
).interactive()
```

```python
small_sample = housing_data.sample(10)
```

```python
# Lets customize using a sample set of data
alt.Chart(small_sample, width=600, height=400).mark_circle(size=60).encode(
    x = 'BuildingArea:Q',
    y='Price:Q',
    color='Regionname:N',
    size='Distance:Q'
).interactive()
```

```python
# Lets customize using a points of data
points = alt.Chart(small_sample, width=600, height=400).mark_circle(size=60).encode(
            x = 'BuildingArea:Q',
            y='Price:Q',
            color='Regionname:N'
            )
points
```

```python
text = points.mark_text(
    align='left',
    baseline='middle',
    dx=8
).encode(
    text='Address'
).interactive()

text
```

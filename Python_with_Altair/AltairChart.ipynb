```python
import altair as alt
import pandas as pd
```

```python
data = pd.DataFrame({'Categories' : ['Clothing',
                                     'Electronics',
                                     'Furniture',
                                     'Health',
                                     'Kitchen'], 
                     'Sales' : [110,343,201,120,278]})

data
```

```python
#Call our new friend altair in this manner -> alt.{method(dataset)}.mark_{graphtype}().encode(x=xaxis,y=yaxis)
#Each chart has a ... that offers save files
alt.Chart(data).mark_bar().encode(x='Categories', y='Sales')
```

```python
# Using height and width now
alt.Chart(
    data,
    height=400,
    width=600
).mark_bar().encode(
    x='Categories',
    y='Sales'
)
```

```python
# Using properties method for the height and width
alt.Chart(data).mark_bar().encode(
    x='Categories',
    y='Sales'
).properties(
    height=400,
    width=600
)
```

```python
# Using title in properties 
alt.Chart(data).mark_bar().encode(
    x='Categories',
    y='Sales'
).properties(
    height=400,
    width=600,
    title='Sales by Product Category'
)
```

```python
# Using using shortform syntax
# O - Original Data (Discrete Data)
# Q - Quantitative Data
alt.Chart(data).mark_bar().encode(
    x='Categories:O',
    y='Sales:Q'
).properties(
    height=400,
    width=600,
    title='Sales by Product Category'
)
```

```python
# This is JSON styled data
jdata = alt.Data(values=[{'Categories' : 'Clothing',  'Sales':110},
                        {'Categories' : 'Electronics',  'Sales':343},
                        {'Categories' : 'Furniture',  'Sales':201},
                        {'Categories' : 'Health',  'Sales':120},
                        {'Categories' : 'Kitchen',  'Sales':278}])

jdata
```

```python
#Let plot the JSON formatted data
alt.Chart(data).mark_bar().encode(
    x='Categories:O',
    y='Sales:Q'
).properties(
    height=400,
    width=600,
    title='Sales by Product Category'
)
```

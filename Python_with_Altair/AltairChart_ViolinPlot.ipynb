```python
import pandas as pd
import altair as alt 
```

```python
# Use the Variables from the box plot script
# pull dataset
auction_data = pd.read_csv('datasets/auction.csv')
# limit rows to 2500
auction_data = auction_data.sample(2500)
# Xbox specific data
auction_xbox = auction_data[auction_data['item'] == 'Xbox game console']
# Watch specific data
auction_watch = auction_data[auction_data['item'] == 'Cartier wristwatch']

```

```python
alt.Chart(auction_watch).transform_density(
    'price',
    as_=['price', 'density']
).mark_area().encode(
    y='density:Q',
    x='price'
).properties(
    width=300
)
```

```python
# Using the stack parameter

alt.Chart(auction_watch).transform_density(
    'price',
    as_=['price', 'density']
).mark_area().encode(
    y=alt.Y('density:Q',stack='center'),
    x='price'
).properties(
    width=300
)
```

```python
# Using horizontal orientation

alt.Chart(auction_watch).transform_density(
    'price',
    as_=['price', 'density']
).mark_area(orient='horizontal').encode(
    x=alt.X('density:Q',stack='center'),
    y='price'
).properties(
    width=300
)
```

```python
# Limiting violin plot range

alt.Chart(auction_watch).transform_density(
    'price',
    as_=['price', 'density'],
    extent=[0,4500]
).mark_area(orient='horizontal').encode(
    x=alt.X('density:Q',stack='center'),
    y='price'
).properties(
    width=300
)
```

```python
# Render for each type of item

alt.Chart(auction_data).transform_density(
    'price',
    as_=['price', 'density'],
    extent=[0,4500],
    groupby=['item']
).mark_area(orient='horizontal').encode(
    x=alt.X('density:Q',stack='center'),
    y='price',
    column='item'
).properties(
    width=300
)
```

```python
# Differentiating data types

alt.Chart(auction_data).transform_density(
    'price',
    as_=['price', 'density'],
    extent=[0,4500],
    groupby=['item']
).mark_area(orient='horizontal').encode(
    x=alt.X('density:Q',stack='center'),
    y='price',
    column='item',
    color='item'
).properties(
    width=300
)
```

```python
# Make visualization more informative 
# Render for each type of item

alt.Chart(auction_data).transform_density(
    'price',
    as_=['price', 'density'],
    extent=[0,4500],
    groupby=['item', 'auction_type']
).mark_area(orient='horizontal').encode(
    x=alt.X('density:Q',stack='center'),
    y='price',
    column='item',
    color='auction_type'
).properties(
    width=200
)
```

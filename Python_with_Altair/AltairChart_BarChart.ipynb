```python
bar_chart = alt.Chart(cars_data, height=400, width=600).mark_bar().encode(
    x='carbody',
    y='median(horsepower):Q'
)
bar_chart
```

```python
# text not bar graph

text = bar_chart.mark_text().encode(
    text='median(horsepower):Q'
)

text
```

```python
bar_chart + text
```

```python
text = bar_chart.mark_text(dy=-5).encode(
    text='median(horsepower):Q'
)

bar_chart + text
```

```python
bar_chart = alt.Chart(cars_data, height=400, width=600).mark_bar().encode(
    x=alt.X('carbody', sort='y'),
    y='median(horsepower):Q'
)

bar_chart
```

```python
#Descending

bar_chart = alt.Chart(cars_data, height=400, width=600).mark_bar().encode(
    x=alt.X('carbody', sort='-y'),
    y='median(horsepower):Q'
)

bar_chart
```

Augmenting and Customizing Bar Chart

```python
car_sample = cars_data.head(10)
car_sample
```

```python
alt.Chart(car_sample, height=400, width=600).mark_bar().encode(
    x='CarName',
    y='price'
)
```

```python
#Condition for colors -- Where price is less than 18k

alt.Chart(car_sample, height=400, width=600).mark_bar().encode(
    x='CarName',
    y='price',
    color=alt.condition(
        alt.datum.price > 18000,
        alt.value('pink'),
        alt.value('steelblue')
    )
)
```

```python
# Flip the chart
hot_bar_chart = alt.Chart(car_sample, height=400, width=600).mark_bar().encode(
    y='CarName',
    x='price',
    color=alt.condition(
        alt.datum.price > 18000,
        alt.value('pink'),
        alt.value('steelblue')
    )
)

hot_bar_chart
```

```python
#Draw indicator of the mean/average price
rule = alt.Chart(cars_data).mark_rule(color='red').encode(
    x='mean(price):Q'
)

hot_bar_chart + rule
```

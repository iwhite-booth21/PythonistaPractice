```python
import pandas as pd
import numpy as mp 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# /telecomm data
t_data = pd.read_csv("Datasets/telecom_data_churn.csv")
t_data.head()
```

```python
# Tranpose the csv data using describe method and .T
t_data.describe().T
```

```python
# Using catplot to get a box and whiskerplot - using y
sns.catplot(y = "Total day charge", kind = 'box', data = t_data, height = 6)
```

```python
# Using catplot to get a box and whiskerplot - using x
sns.catplot(x = "Total day charge", kind = 'box', data = t_data, aspect = 2)
```

```python
# Using catplot to get a box and whiskerplot - using access level plot
f, ax = plt.subplots(figsize = (12, 7)) # Access level plot
sns.boxplot(y = "Total day calls", data = t_data)
```

```python
# We will add in the Churn param for the boxplot
# Left boxplot is churn of 0 and right boxplot is 1
sns.catplot(x = "Churn", y = "Total day charge", kind = "box", data = t_data, aspect = 2);
```

```python
# using the showfliers param on boxplot
f, ax = plt.subplots(figsize = (12, 7))
sns.boxplot(x = 'International plan', y = 'Total intl charge',
           showfliers = False, data = t_data)
```

```python
# Using the palette param
f, ax = plt.subplots(figsize = (12, 7))
sns.boxplot(x = 'International plan', y = 'Total intl charge', hue = 'Churn',
            palette = 'Set2', data = t_data);
```

```python
# Using the linewidth param to make lines more prominent in plot

f, ax = plt.subplots(figsize = (12, 7))
sns.boxplot(x = 'Churn', y = 'Total intl charge', hue = 'International plan',
            palette = 'Set2', linewidth = 3, data = t_data);
```

```python
# Using the catplot, we will now view a boxen plot
sns.catplot(x = 'International plan', y = 'Total intl charge',
           kind = 'boxen', data = t_data, aspect = 3);
```

```python
# Using the col param
sns.catplot(x = 'International plan', y = 'Total intl charge', hue = 'Churn',
           col = 'Voice mail plan', kind = 'box', data = t_data)
```

```python
sns.catplot(x = "International plan", y = "Total intl charge",
           kind = "boxen", data = t_data, aspect = 3)
```

```python
# Lets represent the KDE estimation of the PDF curve - using the violin catplot
sns.catplot(y = "Number vmail messages", kind = "violin", data = t_data, aspect = 2)
```

```python
# Using the x axis
sns.catplot(x = "Customer service calls", kind = "violin", data = t_data, aspect =3)
```

```python
# Lets add in the churn param
sns.catplot(y = "Number vmail messages", x = "Churn", kind = "violin", data = t_data, aspect = 2)
```

```python
# Add detail to the churn
sns.catplot(y = "Customer service calls", x = "Churn", kind = "violin", data = t_data, aspect = 2)
```

```python
# Using a access level function
f , ax = plt.subplots(figsize = (12, 7))
sns.violinplot(x = "Churn", y = "Total intl charge", data = t_data)
```

```python
# Starting with working on multiple plots
f, ax = plt.subplots(figsize =(12, 7))
sns.violinplot(x = "Churn", y = "Total intl charge",
                scale = "count", data = t_data);
```

```python
# Starting with working on multiple plots changing the scale value now
f, ax = plt.subplots(figsize =(12, 7))
sns.violinplot(x = "Churn", y = "Total intl charge",
                scale = "width", data = t_data);
```

```python
# Using the hue param now
sns.catplot(x = "Churn", y = "Total intl calls", hue = "International plan", kind = "violin", data = t_data, aspect = 2);
```

```python
# Using a more detailed approach to do each plot - also using more info on the figure

sns.catplot(x = "Churn", y = "Total intl calls", hue = "International plan", kind = "violin", bw = 0.2, cut = 0, data = t_data, aspect = 2)

plt.title("Bw:0.2")

sns.catplot(x = "Churn", y = "Total intl calls", hue = "International plan", kind = "violin", bw = 0.5, cut = 0, data = t_data, aspect = 2)

plt.title("Bw:0.5")

sns.catplot(x = "Churn", y = "Total intl calls", hue = "International plan", kind = "violin", bw = 1, cut = 0, data = t_data, aspect = 2)

plt.title("Bw:1")

sns.catplot(x = "Churn", y = "Total intl calls", hue = "International plan", kind = "violin", bw = 4, cut = 0, data = t_data, aspect = 2)

plt.title("Bw:4")
```

```python
# Using the col param
sns.catplot(x = "Churn", y = "Total day minutes", col = "Voice mail plan", kind = "violin", data = t_data, aspect = 1.5)

```

```python
# Using the split param
sns.catplot(x = "Churn", y = "Total day minutes", hue = "Voice mail plan", kind = "violin", split = True, data = t_data, aspect = 1.5)
```

```python
# Using the palette and inner params
sns.catplot(x = "Churn", y = "Total day minutes", hue = "Voice mail plan", kind = "violin", inner = "stick", palette = "Set2", split = True, data = t_data, aspect = 3)
```

```python
# Using the palette and inner params - using quartile setting
sns.catplot(x = "Churn", y = "Total day minutes", hue = "Voice mail plan", kind = "violin", inner = "quartiles", palette = "Set2", split = True, data = t_data, aspect = 3)
```

```python
# Using swarmplot
g = sns.catplot(x = "Churn", y = "Total day charge", kind = "violin", inner = None, data = t_data, aspect = 2)

sns.swarmplot(x = "Churn", y = "Total day charge", color = "k", size = 3, data = t_data, ax = g.ax)
```

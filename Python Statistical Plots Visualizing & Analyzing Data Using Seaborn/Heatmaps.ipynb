```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
credit_data = pd.read_csv('Datasets/credit_train.csv')
```

```python
#Datatyping is a consistent fail type - Adding conversion step
cols = ['Credit Score', 'Annual Income', 'Monthly Debt', 'Bankruptcies']

for col in cols:
    credit_data[col] = (
        credit_data[col]
        .astype(str)
        .str.replace(',', '', regex=False)
    )
    credit_data[col] = pd.to_numeric(credit_data[col], errors='coerce')
```

```python
credit_data[['Credit Score', 'Annual Income',
             'Monthly Debt', 'Bankruptcies']].corr()
```

```python
# Introducing the correlation matrix:
corrmat = credit_data.corr(numeric_only=True)
f, ax = plt.subplots(figsize =(12,6))
sns.heatmap(corrmat, cmap = 'Blues')
```

```python
f, ax = plt.subplots(figsize = (12, 6))
sns.heatmap(corrmat, square=True, cmap='Blues')
```

```python
#Tweaking the heatmap
ax = plt.subplots(figsize = (12,6))
sns.heatmap(corrmat, vmax=0.9, vmin=-0.5, square=True, annot=True, fmt ='0.2f', cmap='Blues')
```

```python
#Tweaking the heatmap cont.
ax = plt.subplots(figsize = (12,6))
sns.heatmap(corrmat, vmax=0.9, square=True, annot=True, fmt ='0.2f', cmap='Blues', cbar=False)
```

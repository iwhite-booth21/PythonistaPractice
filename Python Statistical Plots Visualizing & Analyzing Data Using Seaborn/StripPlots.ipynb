```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_data = pd.read_csv('Datasets/Sales_Mart.csv')
sales_data.head()
```

```python
sales_data.shape
```

```python
sales_data.describe()
```

```python
#Strip-plot if kind param not present | default catplot type
sns.catplot(x="Outlet_Type", y="Item_Outlet_Sales", data=sales_data, aspect = 2)
```

```python
#Using the order param
# Order is a list that takes labels from the x variable (In this case Outlet_Type)
sns.catplot(x="Outlet_Type", y="Item_Visibility", 
order = ["Supermarket Type1", "Supermarket Type2", "Supermarket Type3", "Grocery Store"],
data=sales_data, aspect=2)
```

```python
# Lets use a different type of strip plot
sns.catplot(x="Outlet_Size",y="Item_Outlet_Sales", data=sales_data, aspect=2)
```

```python
#Repeating using the access level function 
f,ax = plt.subplots(figsize = (12,6))
sns.stripplot(x="Outlet_Size", y="Item_Outlet_Sales", data=sales_data)
```

```python
# Adding the jitter param | Jitter helps prevent data points overlap
sns.catplot(x="Outlet_Location_Type",y="Item_Outlet_Sales",jitter=False,data=sales_data,aspect=2)
```

```python
# Using ROW and COL with the strip-plot
# Outlet_Size is a categorical value with 3 distinct sizes (small/medium/high)
sns.catplot(x="Outlet_Type", y="Item_Outlet_Sales", row="Outlet_Size", data=sales_data, aspect=2)
```

```python
#Next Video will be on SWARM PLOTS - 10/27/25

# Starting 10/28/25
# Invoking the value_counts method
sales_data['Item_Type'].value_counts() # x[Column Name].value_counts()
```

```python
# Get a listed of selected items
selected_items = ['Fruits and Vegetables', 'Snack Foods', 'Seafood', 'Hard Drinks']

# Filter rows of the original pandas dataframe these items types (comes from the is in method)
selected_items_sales_data = sales_data[sales_data['Item_Type'].isin(selected_items)]

# Invoke the value_Counts on the filtered data
selected_items_sales_data['Item_Type'].value_counts()
```

```python
sns.catplot(x="Item_Type", y="Item_Visibility", kind="swarm", data=selected_items_sales_data, height=6, aspect=2);
```

```python
# Redoing the SWARM plot using the access level function
f,ax = plt.subplots(figsize=(12,6))
sns.swarmplot(x="Item_Type", y="Item_MRP", data=selected_items_sales_data)
```

```python
stock_data_gm['Date'] = pd.to_datetime(stock_data_gm['Date'], format='%Y-%m-%d')
stock_data_gm.dtypes
```

```python
sns.relplot(x='Date', y='Close', kind='line', data=stock_data_gm, aspect=2)
```

```python
#Plotting Multiple Time series lines

sns.relplot(x='Date', y='Open', kind='line', data=stock_data_gm, color='Green', aspect=4)
plt.title('Opening Price')
sns.relplot(x='Date', y='High', kind='line', data=stock_data_gm, color='Orange', aspect=4)
plt.title('Highest Price')
sns.relplot(x='Date', y='Low', kind='line', data=stock_data_gm, color='Purple', aspect=4)
plt.title('Lowest Price')
sns.relplot(x='Date', y='Close', kind='line', data=stock_data_gm, color='Red', aspect=4)
plt.title('Closing Price')
sns.relplot(x='Date', y='Volume', kind='line', data=stock_data_gm, color='Blue', aspect=4)
plt.title('Volume Traded')
```

```python
#Lets use a access level plot
#Will be  useful later on
f,(ax1,ax2,ax3,ax4,ax5) = plt.subplots(5,1, sharex=True, figsize = (12,6))

sns.lineplot(x='Date', y='Open',  data=stock_data_gm, color='Green', label='Opening Price', ax=ax1)
sns.lineplot(x='Date', y='High',  data=stock_data_gm, color='Orange',label='Highest Price', ax=ax2)
sns.lineplot(x='Date', y='Low',  data=stock_data_gm, color='Purple', label='Lowest Price', ax=ax3)
sns.lineplot(x='Date', y='Close', data=stock_data_gm, color='Red', label='Closing Price', ax=ax4)
sns.lineplot(x='Date', y='Volume', data=stock_data_gm, color='Blue', label='Volume Traded', ax=ax5)
```

```python
# Picking back up on 11/8/25

f, ax  = plt.subplots(1,1, figsize = (12,8))

sns.lineplot(x='Date', y='Open',  
             data=stock_data_gm, color='Green', 
             label='Opening Price', ax=ax)
sns.lineplot(x='Date', y='High',  
             data=stock_data_gm, color='Orange', 
             label='Highest Price', ax=ax)
sns.lineplot(x='Date', y='Low',  
             data=stock_data_gm, color='Purple', 
             label='Lowest Price', ax=ax)
sns.lineplot(x='Date', y='Close',  
             data=stock_data_gm, color='Red', 
             label='Closing Price', ax=ax)

plt.legend()
```

```python
#Looking a Date specific data - An idex from the stock_data_gm

stock_data_gm_di = stock_data_gm.set_index('Date')
stock_data_gm_di.head()
```

```python
# Convert Date to datetime format - Ensure the prior section is active
# - stock_data_gm_di['Date'] = pd.to_datetime(stock_data_gm_di['Date'])
# - stock_data_gm_di = stock_data_gm.set_index('Date', inplace=True)
dcol = ['Open','High','Low','Close']
stock_weekly_mean_gm = stock_data_gm_di[dcol].resample('W').mean()
stock_weekly_mean_gm.head()
```

```python
# Adjusting the time for month rather than week - ise ME, M depricated
stock_monthly_mean_gm = stock_data_gm_di[dcol].resample('ME').mean()
stock_monthly_mean_gm.head()
```

```python
# Adjusting to quarterly
stock_quarterly_mean_gm = stock_data_gm_di[dcol].resample('QE').mean()
stock_quarterly_mean_gm.head()
```

```python
# Adjusting to yearly
stock_quarterly_mean_gm = stock_data_gm_di[dcol].resample('YE').mean()
stock_quarterly_mean_gm.head()
```

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
s_data = pd.read_csv('Datasets/Sales_Mart.csv')
s_data.head()
```

```python
s_data.shape
```

```python
s_data.describe()
```

```python
#Strip-plot if kind param not present | default catplot type
sns.catplot(x="Outlet_Type", y="Item_Outlet_Sales", data=s_data, aspect = 2)
```

```python
#Using the order param
# Order is a list that takes labels from the x variable (In this case Outlet_Type)
sns.catplot(x="Outlet_Type", y="Item_Visibility", 
order = ["Supermarket Type1", "Supermarket Type2", "Supermarket Type3", "Grocery Store"],
data=s_data, aspect=2)
```

```python
# Lets use a different type of strip plot
sns.catplot(x="Outlet_Size",y="Item_Outlet_Sales", data=s_data, aspect=2)
```

```python
#Repeating using the access level function 
f,ax = plt.subplots(figsize = (12,6))
sns.stripplot(x="Outlet_Size", y="Item_Outlet_Sales", data=s_data)
```

```python
# Adding the jitter param | Jitter helps prevent data points overlap
sns.catplot(x="Outlet_Location_Type",y="Item_Outlet_Sales",jitter=False,data=s_data,aspect=2)
```

```python
# Using ROW and COL with the strip-plot
# Outlet_Size is a categorical value with 3 distinct sizes (small/medium/high)
sns.catplot(x="Outlet_Type", y="Item_Outlet_Sales", row="Outlet_Size", data=s_data, aspect=2)
```

```python
#Next Video will be on SWARM PLOTS - 10/27/25

# Starting 10/28/25
# Invoking the value_counts method
s_data['Item_Type'].value_counts() # x[Column Name].value_counts()
```

```python
# Get a listed of selected items
selected_items = ['Fruits and Vegetables', 'Snack Foods', 'Seafood', 'Hard Drinks']

# Filter rows of the original pandas dataframe these items types (comes from the is in method)
selected_items_sales_data = s_data[s_data['Item_Type'].isin(selected_items)]

# Invoke the value_Counts on the filtered data
selected_items_sales_data['Item_Type'].value_counts()
```

```python
sns.catplot(x="Item_Type", y="Item_Visibility", kind="swarm", data=selected_items_sales_data, height=6, aspect=2);
```

```python
# Redoing the SWARM plot using the access level function
f,ax = plt.subplots(figsize=(12,6))
sns.swarmplot(x="Item_Type", y="Item_MRP", data=selected_items_sales_data)
```

```python
s_data['Outlet_Establishment_Year'].value_counts()
```

```python
# Get a listed of selected items
selected_years = [1985,1987,1999,2009]

# Filter rows of the original pandas dataframe these items types (comes from the is in method)
selected_years_sales_data = s_data[\
    s_data['Outlet_Establishment_Year'].isin(selected_years)]

# Invoke the value_Counts on the filtered data
selected_years_sales_data['Outlet_Establishment_Year'].value_counts()
```

```python
# Adding the hue param
sns.catplot(x='Outlet_Establishment_Year', y='Item_Outlet_Sales', hue = 'Outlet_Size', kind='swarm', data=selected_years_sales_data, height=8, aspect=2)
```

```python
# Adding the hue param
sns.catplot(x='Outlet_Establishment_Year', y='Item_Outlet_Sales', row = 'Outlet_Size', kind='swarm', data=selected_years_sales_data, height=6, aspect=1.5, dodge= True)
```

```python
g_data = pd.read_csv('Datasets/GM.csv')
g_data.head()
```

```python
g_data.tail()
```

```python
g_data.describe()
```

```python
g_data.dtypes
```

```python
g_data['Date'] = pd.to_datetime(g_data['Date'], format='%Y-%m-%d')
g_data.dtypes
```

```python
sns.relplot(x='Date', y='Close', kind='line', data=g_data, aspect=2)
```

```python
#Plotting Multiple Time series lines

sns.relplot(x='Date', y='Open', kind='line', data=g_data, color='Green', aspect=4)
plt.title('Opening Price')
sns.relplot(x='Date', y='High', kind='line', data=g_data, color='Orange', aspect=4)
plt.title('Highest Price')
sns.relplot(x='Date', y='Low', kind='line', data=g_data, color='Purple', aspect=4)
plt.title('Lowest Price')
sns.relplot(x='Date', y='Close', kind='line', data=g_data, color='Red', aspect=4)
plt.title('Closing Price')
sns.relplot(x='Date', y='Volume', kind='line', data=g_data, color='Blue', aspect=4)
plt.title('Volume Traded')
```

```python
#Lets use a access level plot
#Will be  useful later on
f,(ax1,ax2,ax3,ax4,ax5) = plt.subplots(5,1, sharex=True, figsize = (12,6))

sns.lineplot(x='Date', y='Open',  data=g_data, color='Green', label='Opening Price', ax=ax1)
sns.lineplot(x='Date', y='High',  data=g_data, color='Orange',label='Highest Price', ax=ax2)
sns.lineplot(x='Date', y='Low',  data=g_data, color='Purple', label='Lowest Price', ax=ax3)
sns.lineplot(x='Date', y='Close', data=g_data, color='Red', label='Closing Price', ax=ax4)
sns.lineplot(x='Date', y='Volume', data=g_data, color='Blue', label='Volume Traded', ax=ax5)
```

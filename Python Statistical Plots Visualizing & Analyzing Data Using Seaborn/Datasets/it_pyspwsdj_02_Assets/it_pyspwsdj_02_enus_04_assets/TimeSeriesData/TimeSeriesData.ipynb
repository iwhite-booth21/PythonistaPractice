```python
import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns
```

```python
stock_data_gm = pd.read_csv('Datasets/GM.csv')

stock_data_gm.head()
```

```python
stock_data_gm.dtypes
```

```python
stock_data_gm['Date' ] = pd.to_datetime(stock_data_gm['Date'], format = '%Y-%m-%d')

stock_data_gm.dtypes
```

```python
sns.relplot(x = 'Date', y = 'Close', kind = 'line',
            data = stock_data_gm, aspect = 2)
```


Here,We have drawn lineplots for Daily Opening stock price,Highest stock price, Lowest stock price ,Closing stock price and Volume of GM stocks.The Daily opening,highest,Lowest,Closing stock price curves appears to be similar.
Volume curve is having high fluctuations.Highest volume traded in between 2014 to 2020 was in early 2014 and the value was above 8*10^7.


```python
sns.relplot(x = 'Date', y = 'Open', kind = 'line', 
            data = stock_data_gm , color = 'Green' , aspect = 4)

plt.title('Opening Price')

sns.relplot(x = 'Date', y = 'High', kind = 'line', 
            data = stock_data_gm , color = 'Orange', aspect = 4)

plt.title('Highest Price')

sns.relplot(x = 'Date', y = 'Low', kind = 'line', 
            data = stock_data_gm , color = 'Purple', aspect = 4)

plt.title('Lowest Price')

sns.relplot(x ='Date', y = 'Close', kind = 'line', 
            data = stock_data_gm , color = 'Red', aspect = 4)

plt.title('Closing Price')

sns.relplot(x ='Date', y = 'Volume', kind = 'line',
            data = stock_data_gm , color = 'Blue', aspect = 4)

plt.title('Volume Traded')
```

We have drawn lineplots in Common axes for Daily Opening stock price,Highest stock price, Lowest stock price ,and Closing stock price. In order to resample the data for getting aggregated value of Stock prices, We have set the 'Date' variable as Index(using 'set_index' method in pandas) and formed a new dataframe

```python
stock_data_gm_di = stock_data_gm.set_index('Date')

stock_data_gm_di.head()
```

Firstly,We have resampled the time series data to weekly time series by using 'resample' method in pandas and taken the mean by using 'mean' function.
.The first row above, labelled 2014-01-05, contains the mean of all the data contained in the time bin 2006-01-01 through 2006-01-04.The second row, labelled 2006-01-12, contains the mean data for the 2006-01-05 through 2006-01-11 time bin, and so on.

```python
data_columns = ['Open', 'High', 'Low', 'Close']

stock_weekly_mean_gm = stock_data_gm_di[data_columns].resample('W').mean()

stock_weekly_mean_gm.head()
```

Same way, we have resampled the time series  to Monthly, Quarterly,and Yearly time series

```python
stock_monthly_mean_gm = stock_data_gm_di[data_columns].resample('M').mean()

stock_monthly_mean_gm .head()
```

```python
stock_quarterly_mean_gm = stock_data_gm_di[data_columns].resample('Q').mean()

stock_quarterly_mean_gm .head()
```

```python
stock_yearly_mean_gm = stock_data_gm_di[data_columns].resample('A').mean()

stock_yearly_mean_gm .head(7)
```

Here,We have drawn line plots for Closing Price of the Stock. It can be observed clearly that as the time series data gets aggregated less frequently, the curves become more smooth and fluctuations are getting reduced. The daily time series line plot is having highest fluctuations and Yearly aggregated time series line plot is having lowest fluctuations. In the same way, Other price variables can be drawn and analysed.

```python
fig, (ax1, ax2, ax3, ax4, ax5) = plt.subplots(5, 1, figsize = (12, 8), sharex = True)

sns.lineplot(x = stock_data_gm_di.index, y = 'Close',
             data = stock_data_gm_di, color = 'Green', 
             ax = ax1, label = 'Daily')

sns.lineplot(x = stock_weekly_mean_gm.index, y = 'Close',
             data = stock_weekly_mean_gm, color = 'Red',
             ax = ax2, label = 'Weekly')

sns.lineplot(x = stock_monthly_mean_gm.index, y = 'Close',
             data = stock_monthly_mean_gm, color = 'Blue',
             ax = ax3, label = 'Monthly')

sns.lineplot(x = stock_quarterly_mean_gm.index, y = 'Close',
             data = stock_quarterly_mean_gm, color = 'Magenta', 
             ax = ax4, label = 'Quarterly')

sns.lineplot(x = stock_yearly_mean_gm.index, y = 'Close',
             data = stock_yearly_mean_gm, color = 'Orange',
             ax = ax5, label = 'Yearly')

plt.legend()
```

In order to plot the same aggregated plots in weekly, monthly ,quarterly, yearly using Relplot, we have to reset the index of Dataframes and then pass the 'Date' and Stock price(which in this case is Closing price) to 'x' and 'y' arguments of relplot function.

```python
stock_weekly_mean_gm = stock_weekly_mean_gm.reset_index()

stock_monthly_mean_gm = stock_monthly_mean_gm.reset_index()

stock_quarterly_mean_gm = stock_quarterly_mean_gm.reset_index()

stock_yearly_mean_gm = stock_yearly_mean_gm.reset_index()
```

Here are the same plots using relplot function

```python
sns.relplot(x = 'Date', y = 'Close', kind = 'line', 
            data = stock_data_gm, color = 'Green', aspect = 4)

plt.title('Daily Closing Price')

sns.relplot(x = 'Date', y = 'Close', kind = 'line', 
            data = stock_weekly_mean_gm, color = 'Blue', aspect = 4)

plt.title('Weekly Average Closing Price')

sns.relplot(x = 'Date', y = 'Close', kind = 'line', 
            data = stock_monthly_mean_gm, color = 'Magenta', aspect = 4)

plt.title('Monthly Average Closing Price')

sns.relplot(x = 'Date', y = 'Close', kind = 'line', 
            data = stock_quarterly_mean_gm, color = 'Red', aspect = 4)

plt.title('Quarterly Average Closing Price')

sns.relplot(x = 'Date', y = 'Close', kind = 'line', 
            data = stock_yearly_mean_gm, color = 'Orange', aspect = 4)

plt.title('Yearly Average Closing Price')
```

Extracting 'year','month','quarter' from the dataset and forming new variables and viewing the modified dataset.

```python
stock_data_gm['year'] = stock_data_gm['Date'].dt.year

stock_data_gm['month'] = stock_data_gm['Date'].dt.month

stock_data_gm['quarter'] = stock_data_gm['Date'].dt.quarter

stock_data_gm.sample(5)
```

We have drawn separate  lineplots for each month using hue(month)  with 'Year' on X-axis and 'Closing Price' on Y-axis.
Size of the confidence interval to draw when aggregating with an estimator can be customised using 'ci'. We have disabled confidence interval plot(ci=None) for clear visualisation and customised our hue palette.
Mean value of Closing price for each month in each year is plotted. Mean value for the month of October in the year 2017 is the highest in the period from 2015 to 2020.


```python
f, ax = plt.subplots(figsize = (12, 8))

sns.lineplot(data = stock_data_gm, x = 'year', y = 'Close', 
             hue = 'quarter', ci = None, palette = 'Set2')
```

We can customise  confidence intervals with translucent error bands or discrete error bars by using 'err_style' argument.By default,err_style is 'band'

```python
f, ax = plt.subplots(figsize = (12, 8))

sns.lineplot(data = stock_data_gm, x = 'year', y = 'Close',
             hue = 'month', err_style = 'band', palette = 'Paired')
```

Here,We have set the err_style to 'bars'.So discrete error bars are visible at each Year for each monthly plots 

```python
f, ax = plt.subplots(figsize = (12, 8))

sns.lineplot(data = stock_data_gm, x = 'year', y = 'Close', 
             hue = 'month', err_style = 'bars', palette = 'Paired')
```

Here, We have used ci=“sd”  to draw the standard deviation of the data instead of Confidence interval.

We have added 'quarter' as both  hue and style. Separate plotting styles are shown for each level of  quarter.
We have enabled the markers for different levels of the style variable.

```python
f, ax = plt.subplots(figsize = (12, 8))

sns.lineplot(data = stock_data_gm, x = 'year', y = 'Close', 
             hue = 'quarter', style = 'quarter', markers = True, 
             ci = 'sd', palette = 'Set2')
```
